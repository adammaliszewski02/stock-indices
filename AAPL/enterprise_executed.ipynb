```python
import os

from google import genai
from google.genai import types

# Set GEMINI_API_KEY and STOCK_TICKER in the environment before running this notebook.
symbol = os.environ.get("STOCK_TICKER", "").strip().upper()
if not symbol:
    raise RuntimeError("Set STOCK_TICKER before running enterprise.ipynb")

FALLBACK_DESCRIPTIONS = {
    "PG": "Procter & Gamble (NYSE: PG) is an American multinational consumer goods corporation headquartered in Cincinnati, Ohio. The company specializes in personal care, hygiene, and household cleaning products across five main business segments: Beauty, Grooming, Health Care, Fabric & Home Care, and Baby, Feminine & Family Care. Its portfolio includes Tide, Pampers, Gillette, Crest, and Head & Shoulders. Founded in 1837, P&G serves consumers in more than 180 countries and focuses on essential everyday products, product innovation, and global supply networks.",
    "AAPL": "Apple Inc. (NASDAQ: AAPL) designs, manufactures, and sells smartphones, personal computers, tablets, wearables, and related services. Its ecosystem includes iPhone, Mac, iPad, Apple Watch, AirPods, iOS, macOS, the App Store, and iCloud. Founded in 1976 and headquartered in Cupertino, California, Apple serves customers worldwide through integrated hardware, software, and services.",
}

api_key = os.getenv("GEMINI_API_KEY")
if api_key:
    try:
        client = genai.Client(api_key=api_key)
        description_response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=f"What is the company with ticker {symbol} doing?",
            config=types.GenerateContentConfig(
                system_instruction="You must answer professionally in maximum 200 words. Describe the company in a concise manner. Avoid bullet points. You can include some historical facts.",
                temperature=0.2,
            )
        )
        description = description_response.text.strip()
    except Exception as error:
        print(f"AI description unavailable for {symbol}; using the local profile.")
        description = FALLBACK_DESCRIPTIONS.get(symbol, f"Company profile unavailable for {symbol}.")
else:
    description = FALLBACK_DESCRIPTIONS.get(symbol, f"Company profile unavailable for {symbol}.")

print(f"Company description: {symbol}\n")
print(description)
```
